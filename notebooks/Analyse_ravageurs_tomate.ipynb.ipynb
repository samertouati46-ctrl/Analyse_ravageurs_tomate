```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Bibliothèques chargées")
```

```python
data = {
    "Parcelle":["P1","P2","P3","P4","P5","P6","P7","P8","P9"],
    "Traitement":["T0","T0","T0","T1","T1","T1","T2","T2","T2"],
    "Insectes":[45,52,48,22,18,25,12,15,10],
    "Hauteur_cm":[35,32,34,48,50,47,55,58,56],
    "Rendement_kg":[2.5,2.2,2.4,4.3,4.5,4.1,5.2,5.5,5.1]
}

df = pd.DataFrame(data)

df
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x="Traitement",
    y="Insectes",
    data=df
)

plt.title("Effect of treatments on pest infestation")
plt.show()
```

Interprétation :

Les parcelles non traitées présentent les niveaux d'infestation les plus élevés.
Les traitements T1 et T2 réduisent le nombre d'insectes observés.
